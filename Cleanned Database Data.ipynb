```python
import gc
import re
from pathlib import Path
import polars as pl
import duckdb
```

# TTS DATABASE

```python
path_chinh = r"C:\Users\lamnv5_vtp\Downloads\Streamlit\Tiktok\Phat\TTS_phat_database.parquet"
path_phu = r"C:\Users\lamnv5_vtp\Downloads\DanhGia\Data_FM_MM_LM.parquet"

# Sửa lại path_out phải trỏ đến tận tên file parquet đích
thu_muc_out = r"C:\Users\lamnv5_vtp\Downloads\Database\TTS"
Path(thu_muc_out).mkdir(parents=True, exist_ok=True) # Tự động tạo thư mục nếu chưa có

path_out = str(Path(thu_muc_out) / "TTS_DATABASE.parquet")

duckdb.sql(f"""
    COPY (
        SELECT 
            t1.*, 
            t2.tg_nhap_dt, t2.ma_kh, t2.loai_van_chuyen, t2.tg_lx_xnbg_bcg, 
            t2.tg_nhap_bcp, t2.tg_phat_l1, t2.tg_ptc, t2.tg_tt_pl1, t2.tg_tt_ptc, 
            t2.tg_tt_kpi, t2.danhgia_tt_pl1, t2.danhgia_tt_ptc, t2.tg_fm, 
            t2.tg_fm_kpi, t2.danhgia_fm, t2.tg_mm, t2.tg_mm_kpi, t2.danhgia_mm, 
            t2.tg_lm_pl1, t2.tg_lm_ptc, t2.tg_lm_kpi, t2.danhgia_lm_pl1, t2.danhgia_lm_ptc
        FROM read_parquet('{path_chinh}') t1
        LEFT JOIN read_parquet('{path_phu}') t2 
            ON t1.ma_phieugui = t2.ma_phieugui
    ) TO '{path_out}' (FORMAT PARQUET, COMPRESSION 'ZSTD');
""")

print(f"Đã xuất file thành công tại: {path_out}")
```
